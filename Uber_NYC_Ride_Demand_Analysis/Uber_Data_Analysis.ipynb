### Cleaning and Understanding the data before analysis

```python
df['Date/Time'] = pd.to_datetime(df['Date/Time'],format ="%m/%d/%Y %H:%M:%S")
```

```python
df['Week'] = df['Date/Time'].dt.day_name()
df['Month'] = df['Date/Time'].dt.month_name()
df['Hour_Of_Day'] = df['Date/Time'].dt.hour
df['Min_Of_Day'] = df['Date/Time'].dt.minute
df.head()
```

# Exploration and Visualization

## 1. The number of trips by hour

```python
peak_time= df['Hour_Of_Day'].value_counts().sort_index()
peak_time
```

```python
peak_time.plot(kind = 'line', figsize = (6,6),color = 'red')
plt.title("Number of Trips Per Hour")
plt.xlabel("Trip by Hour")
plt.show()
```

## 2. The number of trips by month

```python
peak_time= df['Hour_Of_Day'].value_counts().head(1)
```

```python
months = ['April','May','June','July','August','September']
df['Month'] = pd.Categorical(df['Month'],categories= months,ordered = True)
trip_by_month = df['Month'].value_counts().sort_index()
trip_by_month
```

```python
trip_by_month.plot(kind='barh', figsize=(10,6), color = 'green')
plt.title("Trips by Month")
plt.xlabel("Number of Trips")
plt.ylabel("Months")
plt.gca().invert_yaxis()
plt.show()
```

**Ratio of increase from August to September**

```python
Number_of_trips_in_August = trip_by_month.loc['August']
Number_of_trips_in_September = trip_by_month.loc['September']
ratio_month =round((Number_of_trips_in_September-Number_of_trips_in_August)*100/Number_of_trips_in_August,0)
print(f"Ratio of increase from August to September {ratio_month}%")
```

## 3. The number of trips by weekday

```python
#Grouping by Week
Week = ['Sunday','Monday','Tuesday','Wednesday','Thursday','Friday','Saturday']
df['Week'] = pd.Categorical(df['Week'],categories= Week,ordered = True)

df_week_grouped = df.groupby('Week', observed = False).count()
#Creating the grouped DataFrame
df_week = pd.DataFrame({'Number_of_trips':df_week_grouped.values[:,0]}, index = df_week_grouped.index) 
df_week
```

```python
df_week.plot(kind='barh', figsize=(10,6), color = 'blue')
plt.title("Trips per day")
plt.xlabel("Number of Trips")
plt.ylabel("Day")
plt.gca().invert_yaxis()
plt.show()
```

## 4. The number of trips by Month and Week

```python
fig = plt.gcf()
fig.set_size_inches(14,6)
sns.catplot(data = df,kind = 'count',x = 'Month', hue = 'Week')
plt.title("Number of Trips by Month and Week")
plt.xlabel("Month")
plt.ylabel("Number of Trips")
plt.xticks(rotation  = 30)
plt.show()
```

```python
#Grouping by Weekday and Month
df_day_minute = df.groupby(['Week','Min_Of_Day']).size().reset_index(name = "Number_Of_Trips")
df_day_minute
```

## 5. Trip Density by Day and Minute of Day

```python
day_minute = df_day_minute.pivot(index='Week',columns='Min_Of_Day',).fillna(0)
days = ['Sunday', 'Monday', 'Tuesday','Wednesday', 'Thursday', 'Friday', 'Saturday']
day_minute = day_minute.reindex(days)
plt.figure(figsize=(18, 6))
sns.heatmap(day_minute,cmap='YlOrRd')
plt.title('Trip Density by Day and Minute of Day')
plt.xlabel('Minute of Day')
plt.ylabel('Day')
plt.tight_layout()
plt.show()
```

## 6. Pickup Density Across New York City

```python
plt.figure(figsize=(10, 8))
plt.hexbin(df['Lon'],df['Lat'],gridsize=50,mincnt=1,cmap='viridis')
plt.colorbar(label='Number of Pickups')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Pickup Density Across New York City')
plt.show()
```

## 7. Trips to different bases by month

```python
base_month = df.groupby(['Month', 'Base']).size().reset_index(name='Trips')
plt.figure(figsize=(12,6))
sns.barplot(data=base_month,x='Month',y='Trips',hue='Base')
plt.title('Trips by Base and Month')
plt.xlabel('Month')
plt.ylabel('Number of Trips')
plt.xticks(rotation=45)
plt.legend(title='Base')
plt.tight_layout()
plt.show()
```

## 8. Trip density of different bases per hour

```python
base_hour = df.pivot_table(index='Base',columns='Hour_Of_Day',aggfunc='size', fill_value = 0)

plt.figure(figsize=(14,6))

sns.heatmap(base_hour,cmap='YlOrRd')

plt.title('Trip Density by Base and Hour')
plt.xlabel('Hour of Day')
plt.ylabel('Base')

plt.tight_layout()
plt.show()
```

## 9. Trip Densisty by day and hour heatmap

```python
day_hour = df.pivot_table(index='Week',columns='Hour_Of_Day',aggfunc='size',fill_value=0)
days = ['Monday', 'Tuesday', 'Wednesday','Thursday', 'Friday', 'Saturday', 'Sunday']
day_hour = day_hour.reindex(days)
plt.figure(figsize=(16, 6))
sns.heatmap(day_hour,cmap='YlOrRd')
plt.title('Trip Density by Day and Hour')
plt.xlabel('Hour of Day')
plt.ylabel('Day of Week')
plt.tight_layout()
plt.show()
```

## 10. Comparison of trips in weekdays and weekends

```python
df['Day_Type'] = df['Week'].apply(lambda x: 'Weekend' if x in ['Saturday', 'Sunday'] else 'Weekday')
hour_comparison['Pct'] = hour_comparison.groupby('Day_Type')['Trips'].transform(lambda x: x / x.sum() * 100)
plt.figure(figsize=(14, 6))
sns.lineplot(data=hour_comparison, x='Hour_Of_Day', y='Pct', hue='Day_Type', marker='o')
plt.title('Weekday vs Weekend Trips by Hour')
plt.xlabel('Hour of Day')
plt.ylabel('Number of Trips')
plt.xticks(range(24))
plt.tight_layout()
plt.show()
```

## 11. Calculate Month-over-Month growth

```python
monthly_trips = df.groupby('Month').size().reset_index(name='Trips')
monthly_trips['MoM_Growth_%'] = monthly_trips['Trips'].pct_change() * 100
monthly_trips
```

```python
plt.figure(figsize=(10, 5))
sns.lineplot(data=monthly_trips,x='Month',y='MoM_Growth_%',marker='o')
plt.title('Month-over-Month Trip Growth (April to September)')
plt.xlabel('Month')
plt.ylabel('MoM Growth (%)')
plt.axhline(0, linestyle='--')
plt.tight_layout()
plt.show()
```

# Conclusion and Inferences
- Overall demand grew steadily. Trips rose every month from April (556,767) to September (1,004,099) — nearly doubling over six months, with the biggest single jump being a 23% increase from August to September alone.
- Weekdays are busier than weekends. Thursday (741,372) and Wednesday (683,604) saw the most trips, while Sunday (481,158) was the quietest day — suggesting Uber usage in this period was driven more by weekday commuting and work-related travel than weekend leisure trips.
- Trip activity peaks in the evening. The highest trip volumes cluster around the early evening hours (roughly 6–9 PM), consistent with an evening commute and after-work travel pattern.
- Weekday and weekend usage follow different hourly rhythms. The weekday-vs-weekend hourly comparison shows the two groups don't peak at the same times weekdays show sharper commute-driven spikes, while weekends are more spread out later in the day.
- Pickups are heavily concentrated in specific areas of the city, not spread evenly across New York — the pickup density map shows clear hotspots rather than uniform coverage citywide.
- Different dispatch bases behave differently. Trip volume isn't split evenly across the 5 bases — some clearly handle more volume than others, and their activity by hour and by month isn't identical, suggesting each base may serve a different segment of demand.
- Data quality needed real attention. The raw dataset contained 82,581 duplicate rows and one redundant column (Day, which just repeated Week) — both were identified and cleaned before analysis, which was necessary to avoid inflating trip counts.
- Bottom line: Uber demand in NYC during this period was driven primarily by weekday, evening, commute-style travel, concentrated in specific city zones and unevenly distributed across dispatch bases — with clear month-over-month growth throughout the observed period.
